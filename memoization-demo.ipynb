```python
# fib

def fib(n):
    if n==1 or n==0:
        return 1
    else:
        return fib(n-1) + fib(n-2)
```

```python
import time

start = time.time()
a = fib(38)
print(a, time.time() - start)
```

```python
# Applying memoization in fib function

def fibo(n, d):
    if n in d:
        return d[n]
    else:
        d[n] = fibo(n-1, d) + fibo(n-2, d)
        return d[n]
```

```python
start_new = time.time()
d = {
    0:1,
    1:1
}
b = fibo(38,d)
print(b, time.time()-start_new)
```
